```python
import pandas as pd
```

```python
data=pd.read_csv("test.csv")
data
```

```python
p=[]
for i in range(0,533):
  testvalue = {
     "Temperature" : data["Temperature"][i],
      "Humidity" : data["Humidity"][i],
     "Light" : data["Light"][i],
     "CO2" :  data["CO2"][i],
      "HR" : data["HR"][i],
     }

  response = project.classify(testvalue)
  top_match = response[0]

  label = int(top_match["class_name"])
  p.append(label)
```

```python
data.rename(columns={"Occupancy": "truth"}, inplace=True)
```

```python
data["prediction"]=p

data
```

```python
data.to_excel("predictions.xlsx",index=False)
```

```python
import pandas as pd
```

```python
data=pd.read_excel("predictions.xlsx")
data
```

```python
from sklearn.metrics import confusion_matrix , accuracy_score , precision_score
import matplotlib.pyplot  as plt
import seaborn as sns
```

```python
cm=confusion_matrix(data["truth"],data["prediction"])

plt.figure(figsize=(6,5))

sns.heatmap(cm,
            annot=True,
            fmt="d",
            cmap="mako")

plt.xlabel("prediction")
plt.ylabel("truth")
plt.show()
```

```python
accuracy_score(data["truth"],data["prediction"])
```

```python
precision_score(data["truth"],data["prediction"],pos_label=0),precision_score(data["truth"],data["prediction"],pos_label=1)
```
